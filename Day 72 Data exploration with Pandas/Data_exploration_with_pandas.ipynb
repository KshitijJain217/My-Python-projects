* Use .head(), .tail(), .shape and .columns to explore your DataFrame and
find out the number of rows and columns as well as the column names.

* Look for NaN (not a number) values with .findna() and consider using .dropna() to clean up your DataFrame.

* You can access entire columns of a DataFrame using the square bracket notation: df['column name'] or df[['column name 1', 'column name 2', 'column name 3']]

* You can access individual cells in a DataFrame by chaining square brackets df['column name'][index] or using df['column name'].loc[index]

* The largest and smallest values, as well as their positions, can be found with methods like .max(), .min(), .idxmax() and .idxmin()

* You can sort the DataFrame with .sort_values() and add new columns with .insert()

* To create an Excel Style Pivot Table by grouping entries that belong to a particular category use the .groupby() method

### College Major with the Highest Mid-Career Salary

```python
highest_mid_career_salary = clean_df['Mid-Career Median Salary'].max()
highest_mid_career_major_idx = clean_df['Mid-Career Median Salary'].idxmax()
highest_mid_career_major = clean_df['Undergraduate Major'].loc[highest_mid_career_major_idx]

print(f"The college major with the highest mid-career salary is {highest_mid_career_major} with a median salary of ${highest_mid_career_salary:,.2f}.")
```

### College Major with the Lowest Starting Salary

```python
lowest_starting_salary = clean_df['Starting Median Salary'].min()
lowest_starting_major_idx = clean_df['Starting Median Salary'].idxmin()
lowest_starting_major = clean_df['Undergraduate Major'].loc[lowest_starting_major_idx]

print(f"The college major with the lowest starting salary is {lowest_starting_major} with a median salary of ${lowest_starting_salary:,.2f}.")
```

### College Major with the Lowest Mid-Career Salary

```python
lowest_mid_career_salary = clean_df['Mid-Career Median Salary'].min()
lowest_mid_career_major_idx = clean_df['Mid-Career Median Salary'].idxmin()
lowest_mid_career_major = clean_df['Undergraduate Major'].loc[lowest_mid_career_major_idx]

print(f"The college major with the lowest mid-career salary is {lowest_mid_career_major} with a median salary of ${lowest_mid_career_salary:,.2f}.")
```

```python
clean_df.loc[clean_df['Mid-Career Median Salary'].idxmin()]
```

```python
spread_col = clean_df['Mid-Career 90th Percentile Salary']
clean_df.insert(1, 'Spread', spread_col)
clean_df.head()
```

```python
low_risk = clean_df.sort_values('Spread')
low_risk[['Undergraduate Major', 'Spread']].head()
```

```python
# Majors with the Highest Potential
highest_potential = clean_df.sort_values('Mid-Career 90th Percentile Salary', ascending = False)
highest_potential[['Undergraduate Major', 'Mid-Career 90th Percentile Salary']].head()
```

```python
# Majors with the Greatest Spread in Salaries
highest_spread = clean_df.sort_values('Spread', ascending=False)
highest_spread[['Undergraduate Major', 'Spread']].head()
```

```python
clean_df.groupby('Group').count()
```

```python
clean_df.groupby('Group')[['Starting Median Salary', 'Mid-Career Median Salary', 'Mid-Career 10th Percentile Salary', 'Mid-Career 90th Percentile Salary']].mean()
```
